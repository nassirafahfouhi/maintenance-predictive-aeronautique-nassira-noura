```python
import pandas as pd
import numpy as np

# -------------------------------------------------------------------
# Construction de la liste des noms de colonnes du dataset FD001
# -------------------------------------------------------------------
# Le fichier brut NASA ne contient pas d'en-têtes.
# On recrée donc explicitement :
# - 'engine_id' : identifiant unique du moteur
# - 'cycle'     : numéro de cycle (temps discret)
# - 'setting_1..3' : paramètres opérationnels (profil de mission)
# - 's1..s21'      : 21 capteurs (températures, pressions, vitesses, etc.)
cols = ["engine_id", "cycle"] + \
       [f"setting_{i}" for i in range(1, 4)] + \
       [f"s{i}" for i in range(1, 22)]

# -------------------------------------------------------------------
# Chargement des fichiers C-MAPSS FD001
# -------------------------------------------------------------------
# train_FD001.txt : historique complet de 100 moteurs jusqu'à la panne.
# test_FD001.txt  : 100 autres moteurs observés jusqu'à un certain cycle.
# RUL_FD001.txt   : RUL vraie (en cycles restants) au dernier cycle de
#                   chaque moteur du jeu de test.
# Le séparateur est un ou plusieurs espaces (regex \s+), il n'y a pas d'en-tête,
# on applique donc la liste 'cols' pour nommer les colonnes.
train = pd.read_csv("train_FD001.txt",
                    sep=r"\s+", header=None, names=cols)
test = pd.read_csv("test_FD001.txt",
                   sep=r"\s+", header=None, names=cols)
rul = pd.read_csv("RUL_FD001.txt",
                  sep=r"\s+", header=None, names=["RUL"])

# -------------------------------------------------------------------
# Vérification rapide du chargement
# -------------------------------------------------------------------
# - train.head() : aperçu des premières lignes pour valider la structure.
# - train.shape  : dimensions du jeu d'entraînement (nb lignes, nb colonnes).
# - test.shape   : dimensions du jeu de test.
# - rul.head()   : premières valeurs de RUL associées aux moteurs de test.
train.head(), train.shape, test.shape, rul.head()
```

```python
# -------------------------------------------------------------------
# Aperçu rapide de la structure des jeux de données
# -------------------------------------------------------------------

# Dimensions des jeux :
# - train.shape : nb de lignes / colonnes pour le jeu d'entraînement
# - test.shape  : nb de lignes / colonnes pour le jeu de test
# - rul.shape   : nb de moteurs dans le fichier RUL de test
print("train shape:", train.shape)
print("test shape :", test.shape)
print("rul shape  :", rul.shape)

# Liste complète des colonnes disponibles dans le train.
# Permet de vérifier que la reconstruction des noms (engine_id, cycle,
# settings, capteurs) est correcte.
print("\nColonnes :")
print(train.columns.tolist())

# Vérification du nombre de moteurs simulés dans chaque jeu.
# On s'assure ici que FD001 contient bien 100 moteurs en train et 100 en test.
print("\nNombre d'engins dans train :", train["engine_id"].nunique())
print("Nombre d'engins dans test  :", test["engine_id"].nunique())

# Plage de cycles observée pour les premiers moteurs du train.
# Objectif : confirmer que chaque moteur commence bien au cycle 1
# et qu'il est suivi jusqu'à sa panne (cycle max variable d'un moteur à l'autre).
print("\nCycles min / max par engin (train) :")
print(train.groupby("engine_id")["cycle"].agg(["min", "max"]).head())
```

```python
# Colonnes capteurs
sensor_cols = [c for c in train.columns if c.startswith("s")]

# Stats descriptives des capteurs
train[sensor_cols].describe().T
```

```python
# -------------------------------------------------------------------
# Identification des capteurs constants (variance nulle)
# -------------------------------------------------------------------
# Objectif :
# repérer les capteurs dont la valeur ne varie jamais dans tout le
# jeu d'entraînement. Un capteur avec une seule valeur unique
# (nunique() == 1) ne porte aucune information sur l'état de santé
# du moteur : il ne permet pas de distinguer un moteur neuf d'un
# moteur dégradé.
# Ces capteurs seront donc supprimés des features pour :
# - réduire le bruit et la dimension du problème,
# - accélérer l'entraînement des modèles,
# - éviter d'introduire des variables inutiles dans la Random Forest.

const_sensors = []
for c in sensor_cols:
    if train[c].nunique() == 1:
        const_sensors.append(c)

print("Capteurs constants :", const_sensors)
```

```python
import matplotlib.pyplot as plt

# -------------------------------------------------------------------
# Visualisation de l'évolution de quelques capteurs dans le temps
# -------------------------------------------------------------------
# On se focalise sur un moteur donné (engine_id = 1) pour analyser
# qualitativement l'évolution de ses capteurs en fonction du nombre
# de cycles.
# Cette étape d'EDA permet :
# - de vérifier que certains capteurs dérivent avec l'usure (tendance),
# - d'identifier ceux qui restent presque constants,
# - de relier visuellement la dégradation physique du moteur à ses mesures.

engine_id = 1
# Sélection des lignes correspondant au moteur 1 et tri par numéro de cycle
df1 = train[train["engine_id"] == engine_id].sort_values("cycle")

plt.figure(figsize=(10, 6))

# On trace ici quelques capteurs représentatifs (s2, s3, s4, s7, s9)
# pour observer leur comportement au cours de la vie du moteur.
for c in ["s2", "s3", "s4", "s7", "s9"]:
    plt.plot(df1["cycle"], df1[c], label=c)

plt.xlabel("cycle")
plt.ylabel("valeur capteur")
plt.title(f"Evolution capteurs sélectionnés - moteur {engine_id}")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# -------------------------------------------------------------------
# Sélection des variables pertinentes pour la prédiction de la RUL
# -------------------------------------------------------------------
# À partir de l'analyse descriptive et des capteurs constants, on a
# identifié plusieurs colonnes non informatives :
#   - setting_3 : paramètre opérationnel constant dans FD001
#   - s1, s5, s10, s16, s18, s19 : capteurs dont la valeur ne varie
#     quasiment pas (variance nulle ou très proche de zéro)
# Ces variables n'apportent aucune information sur la dégradation du
# moteur et risquent :
#   - d'ajouter du bruit inutile,
#   - d'augmenter le temps de calcul,
#   - de complexifier le modèle sans gain de performance.
useless_cols = ["setting_3", "s1", "s5", "s10", "s16", "s18", "s19"]

# On construit ensuite la liste finale des features utilisées pour
# l'apprentissage :
#   - on retire 'engine_id' (identifiant purement administratif,
#     non généralisable à de nouveaux moteurs),
#   - on retire 'cycle' pour éviter de donner directement au modèle
#     l'information de temps brut (qui peut créer une dépendance
#     artificielle ou une "fuite" vers la RUL),
#   - on retire toutes les colonnes inutiles définies dans useless_cols.
feature_cols = [
    c for c in train.columns
    if c not in ["engine_id", "cycle"] + useless_cols
]

feature_cols
```

```python
# -------------------------------------------------------------------
# Calcul de la Remaining Useful Life (RUL) pour chaque moteur
# -------------------------------------------------------------------
# Objectif :
# transformer le problème en régression supervisée en associant à
# chaque ligne (moteur, cycle) une RUL = nombre de cycles restants
# avant la panne simulée.

# 1) Cycle maximal (panne) pour chaque moteur
# On regroupe les données par moteur et on récupère, pour chacun,
# le dernier cycle observé. Ce cycle correspond à la panne simulée.
max_cycle_per_engine = train.groupby("engine_id")["cycle"].max()

# 2) Rattacher ce cycle maximal à chaque ligne du moteur
# On crée une nouvelle colonne 'max_cycle' contenant, pour chaque
# ligne, le cycle de panne associé au moteur correspondant.
train = train.merge(
    max_cycle_per_engine.rename("max_cycle"),
    on="engine_id",
    how="left"
)

# 3) Calcul de la RUL
# RUL = max_cycle - cycle courant
# -> plus le moteur avance dans le temps, plus la RUL diminue jusqu'à 0.
train["RUL"] = train["max_cycle"] - train["cycle"]

# 4) Nettoyage
# La colonne 'max_cycle' n'est plus utile une fois la RUL calculée.
# On la supprime pour ne pas l'utiliser comme feature et éviter
# toute fuite d'information vers la cible.
train.drop(columns=["max_cycle"], inplace=True)

# 5) Vérification rapide
# On affiche quelques lignes (engine_id, cycle, RUL) ainsi que
# les statistiques descriptives de la cible RUL (min, max, moyenne, etc.)
train[["engine_id", "cycle", "RUL"]].head(), train["RUL"].describe()
```

```python
import matplotlib.pyplot as plt

# -------------------------------------------------------------------
# Visualisation de la distribution de la cible RUL
# -------------------------------------------------------------------
# Objectif :
# analyser comment se répartissent les valeurs de RUL dans le jeu
# d'entraînement :
# - vérifier l'étendue (RUL minimale, maximale),
# - voir si la base est équilibrée ou concentrée sur des RUL faibles,
#   ce qui peut biaiser l'apprentissage du modèle.

plt.figure(figsize=(8, 4))
plt.hist(train["RUL"], bins=50, edgecolor="k")
plt.xlabel("RUL")
plt.ylabel("Nombre d'échantillons")
plt.title("Distribution de la RUL dans le train")
plt.grid(True)
plt.show()
```

```python
# -------------------------------------------------------------------
# Construction de la matrice de features et de la cible
# -------------------------------------------------------------------
# X_train : contient toutes les variables explicatives retenues
#           (capteurs + settings utiles) définies dans 'feature_cols'
#           après la phase de sélection/filtrage.
# y_train : contient la variable cible à prédire, la Remaining Useful Life (RUL).

X_train = train[feature_cols]
y_train = train["RUL"]

# Vérification des dimensions :
# - X_train : nb d'échantillons x nb de features
# - y_train : nb d'échantillons (doit être cohérent avec X_train)
X_train.shape, y_train.shape
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
import numpy as np

# -------------------------------------------------------------------
# Séparation des données en train et validation
# -------------------------------------------------------------------
# X_train : matrice de features (capteurs + settings sélectionnés)
# y_train : cible RUL associée à chaque ligne.
# On garde 20 % des données pour la validation afin d'estimer la
# capacité de généralisation du modèle sur des moteurs/cycles
# jamais vus pendant l'entraînement.
X_tr, X_val, y_tr, y_val = train_test_split(
    X_train, y_train, test_size=0.2, random_state=42
)

# -------------------------------------------------------------------
# Modèle avancé : Random Forest Regressor (baseline)
# -------------------------------------------------------------------
# Choix du modèle :
# - gère naturellement les relations non linéaires entre capteurs et RUL,
# - robuste au bruit et aux interactions entre variables,
# - peu sensible à la mise à l'échelle des features (pas de normalisation).
# Hyperparamètres :
# - n_estimators=200 : nombre d'arbres dans la forêt (compromis stabilité/temps),
# - max_depth=None  : profondeur non limitée (baseline, risque d'overfit),
# - n_jobs=-1       : utilisation de tous les cœurs CPU disponibles,
# - random_state=42 : reproductibilité des résultats.
rf = RandomForestRegressor(
    n_estimators=200,
    max_depth=None,
    n_jobs=-1,
    random_state=42
)

# Entraînement du modèle sur le sous-ensemble d'apprentissage
rf.fit(X_tr, y_tr)

# -------------------------------------------------------------------
# Évaluation sur le jeu de validation
# -------------------------------------------------------------------
# On prédit la RUL pour les échantillons de validation puis on calcule
# la RMSE (Root Mean Squared Error), qui mesure l'erreur moyenne en
# nombre de cycles entre la RUL vraie et la RUL prédite.
y_pred = rf.predict(X_val)
rmse = np.sqrt(mean_squared_error(y_val, y_pred))

rmse  # valeur utilisée dans le rapport comme performance du modèle baseline
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor

# Grille d'hyperparamètres à tester pour la Random Forest
param_grid = {
    "n_estimators": [100, 200],      # Nombre d'arbres (compromis perf / temps)
    "max_depth": [None, 20, 40],     # Profondeur max pour contrôler l'overfitting
    "min_samples_leaf": [1, 3, 5]    # Nb min d'échantillons par feuille (régularisation)
}

# Modèle de base RandomForestRegressor
# Choisi car robuste, gère bien la non-linéarité et ne demande pas de scaling
rf_base = RandomForestRegressor(
    n_jobs=-1,       # Utilise tous les cœurs pour accélérer
    random_state=42  # Pour des résultats reproductibles
)

# GridSearchCV pour trouver la meilleure combinaison d'hyperparamètres
grid = GridSearchCV(
    rf_base,
    param_grid=param_grid,
    cv=3,                                   # Validation croisée 3-fold
    scoring="neg_root_mean_squared_error",  # On minimise la RMSE
    n_jobs=-1,                              # Calcul parallèle
    verbose=1                               # Affiche la progression
)

# Entraînement avec recherche de la meilleure configuration
grid.fit(X_train, y_train)

# Affiche les meilleurs hyperparamètres et le meilleur score (RMSE positif)
grid.best_params_, -grid.best_score_
```

```python
rf_best = RandomForestRegressor(
    n_estimators=200,
    max_depth=20,
    min_samples_leaf=5,
    n_jobs=-1,
    random_state=42
)

# On réentraîne la Random Forest avec les meilleurs hyperparamètres trouvés (modèle "best").
rf_best.fit(X_tr, y_tr)

# Prédictions sur le jeu de validation avec le nouveau modèle optimisé.
y_pred_best = rf_best.predict(X_val)

# Calcul de la RMSE pour comparer l'ancien modèle (base) et le modèle optimisé (best),
# la RMSE étant la métrique demandée pour évaluer la qualité des prédictions de RUL [file:1].
rmse_best = np.sqrt(mean_squared_error(y_val, y_pred_best))
rmse_base = np.sqrt(mean_squared_error(y_val, y_pred))  # ancien modèle

print("RMSE base  :", rmse_base)
print("RMSE best  :", rmse_best)
```

```python
# On sauvegarde explicitement le meilleur modèle et sa RMSE pour les utiliser plus tard
# (comparaison finale, rapport, test sur FD001, etc.) [file:1].
best_model = rf_best
best_rmse = rmse_best

best_rmse
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Importances des features extraites du meilleur modèle Random Forest.
importances = best_model.feature_importances_

# DataFrame trié pour visualiser les capteurs les plus importants pour la prédiction de la RUL [file:1].
feat_imp = pd.DataFrame({
    "feature": feature_cols,
    "importance": importances
}).sort_values("importance", ascending=False)

feat_imp.head(20)
```

```python
top_k = 15  # Nombre de features les plus importantes à afficher

plt.figure(figsize=(8, 6))  # Taille de la figure
plt.barh(
    feat_imp["feature"].head(top_k)[::-1],      # Noms des features (inversés pour avoir la plus importante en haut)
    feat_imp["importance"].head(top_k)[::-1]    # Valeurs d'importance correspondantes
)
plt.xlabel("Importance (Random Forest)")  # Légende de l'axe X
plt.title("Top features importantes pour la prédiction de la RUL")  # Titre du graphique
plt.grid(axis="x", alpha=0.3)  # Grille légère sur l'axe X pour la lisibilité
plt.tight_layout()  # Ajuste les marges
plt.show()  # Affiche le graphique
```

```python
# Créer un dataframe avec les vraies RUL et prédictions de validation
results_df = X_val.copy()
results_df["engine_id"] = X_val["engine_id"] if "engine_id" in X_val.columns else train.loc[X_val.index, "engine_id"]
results_df["cycle"] = X_val["cycle"] if "cycle" in X_val.columns else train.loc[X_val.index, "cycle"]
results_df["RUL_true"] = y_val.values
results_df["RUL_pred"] = y_pred_best

# Sélectionner quelques moteurs uniques
selected_engines = results_df["engine_id"].unique()[:4]

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

for idx, engine_id in enumerate(selected_engines):
    engine_data = results_df[results_df["engine_id"] == engine_id].sort_values("cycle")
    
    ax = axes[idx]
    ax.plot(engine_data["cycle"], engine_data["RUL_true"], 'o-', label="RUL True", linewidth=2, markersize=4)
    ax.plot(engine_data["cycle"], engine_data["RUL_pred"], 's--', label="RUL Predicted", linewidth=2, markersize=4)
    ax.set_xlabel("Cycle")
    ax.set_ylabel("RUL (cycles)")
    ax.set_title(f"Engine {engine_id}: True vs Predicted RUL")
    ax.legend()
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
